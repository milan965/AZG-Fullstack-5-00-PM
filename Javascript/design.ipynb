```python
import turtle

screen = turtle.Screen()
screen.bgcolor("black")

t = turtle.Turtle()
t.speed(0)
t.width(2)

colors = [
    "red",
    "orange",
    "yellow",
    "lime",
    "cyan",
    "blue",
    "purple",
    "magenta"
]

# Mandala petals
for i in range(36):

    t.pencolor(colors[i % len(colors)])

    # Outer petals
    for j in range(2):
        t.circle(120, 60)
        t.left(120)
        t.circle(120, 60)
        t.left(120)

    t.left(10)


# Inner circle
t.penup()
t.goto(0, -40)
t.pendown()

t.pencolor("white")
t.circle(40)

# Center flower
for i in range(12):
    t.pencolor(colors[i % len(colors)])
    t.circle(40)
    t.left(30)

t.hideturtle()

turtle.done()
```
